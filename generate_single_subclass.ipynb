```python
import rdflib
from rdflib import RDFS, URIRef

def keep_single_subclass(original, subclass, superclass):
    keep = (URIRef(subclass), RDFS.subClassOf, URIRef(superclass))

    if keep not in original:
        raise ValueError("Inclusione non asserita nell'ontologia.")

    filtered = rdflib.Graph()
    for prefix, namespace in original.namespaces():
        filtered.bind(prefix, namespace)

    for triple in original:
        if triple[1] != RDFS.subClassOf or triple == keep:
            filtered.add(triple)

    return filtered

original = rdflib.Graph().parse(
    "univ-bench-inst.ttl", format="turtle"
)

inclusions = sorted(
    (sub, sup)
    for sub, sup in original.subject_objects(RDFS.subClassOf)
    if isinstance(sub, URIRef) and isinstance(sup, URIRef)
)

for sub, sup in inclusions:
    variant = keep_single_subclass(original, sub, sup)
    # Qui potrai verificare o salvare la variante.
```
